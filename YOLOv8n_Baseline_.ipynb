# Training & Evaluasi YOLOv8n — Baseline (Tanpa Optimasi)

**Skripsi:** Perbandingan Model YOLOv8 dan EfficientDet untuk Object Counting pada Smart Classroom

- Tidak ada hyperparameter tuning otomatis
- Tidak ada modifikasi LR Warmup (memakai nilai default Ultralytics: `warmup_epochs=3.0`, dst — tidak diubah/di-override)
- Tidak ada freeze-backbone fine-tuning
- Tidak ada Test-Time Augmentation (TTA)
- Tidak ada threshold tuning (confidence/IoU pakai default: conf=0.25, iou=0.7)

**Catatan soal augmentasi:** parameter augmentasi bawaan YOLOv8 (mosaic, hsv, flip, dst) tetap **dimatikan** di notebook ini — ini BUKAN bagian dari "optimasi", melainkan keharusan teknis karena dataset sudah dipreprocessing & diaugmentasi di Roboflow (resize 640x640, flip horizontal, rotasi ±5°, brightness/exposure ±10%). Mematikannya mencegah *double augmentation*, bukan untuk meningkatkan akurasi.

## 7. Visualisasi Contoh Data Training (Sanity Check)


```python
def draw_yolo_boxes(image_path, label_path, class_names):
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img.shape[:2]
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                parts = line.strip().split()
                if len(parts) < 5:
                    continue
                cls_id = int(parts[0])
                xc, yc, bw, bh = map(float, parts[1:5])
                x1 = int((xc - bw / 2) * w)
                y1 = int((yc - bh / 2) * h)
                x2 = int((xc + bw / 2) * w)
                y2 = int((yc + bh / 2) * h)
                color = (255, 0, 0) if cls_id == 0 else (0, 255, 0)
                cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
                cv2.putText(img, class_names[cls_id], (x1, max(y1 - 5, 0)),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
    return img

train_img_dir = os.path.join(TRAIN_DIR, 'images')
train_lbl_dir = os.path.join(TRAIN_DIR, 'labels')
sample_files = random.sample(os.listdir(train_img_dir), 4)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, fname in zip(axes, sample_files):
    img_path = os.path.join(train_img_dir, fname)
    lbl_path = os.path.join(train_lbl_dir, os.path.splitext(fname)[0] + '.txt')
    img = draw_yolo_boxes(img_path, lbl_path, CLASS_NAMES)
    ax.imshow(img)
    ax.set_title(fname, fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 12 Simulasi Counting dengan Visualisasi

Mensimulasikan output sistem counting pada beberapa gambar test secara acak: bounding box + panel overlay jumlah objek (gaya dashboard smart classroom), dibandingkan dengan ground truth. Judul gambar berwarna hijau jika jumlah prediksi persis sama dengan ground truth, merah jika ada selisih.

```python
import os
import cv2
import matplotlib.pyplot as plt
import random

# ============================================================
# FALLBACK KONFIGURASI
# ============================================================
try:
    CONF_THRESHOLD, IOU_THRESHOLD
except NameError:
    CONF_THRESHOLD, IOU_THRESHOLD = 0.30, 0.50

try:
    test_img_dir, test_labels_dir
except NameError:
    test_img_dir = os.path.join(TEST_DIR, 'images')
    test_labels_dir = os.path.join(TEST_DIR, 'labels')

try:
    count_ground_truth
except NameError:
    def count_ground_truth(label_path, class_names):
        counts = {name: 0 for name in class_names}

        if os.path.exists(label_path):
            with open(label_path, 'r') as f:
                for line in f:
                    line = line.strip()

                    if not line:
                        continue

                    cls_id = int(line.split()[0])

                    if 0 <= cls_id < len(class_names):
                        counts[class_names[cls_id]] += 1

        return counts


# ============================================================
# FUNGSI VISUALISASI PREDIKSI
# ============================================================
def simulate_counting_visual(
    model,
    img_path,
    label_path,
    class_names,
    conf,
    iou
):

    # Baca gambar
    img_bgr = cv2.imread(img_path)

    if img_bgr is None:
        raise ValueError(f"Gambar tidak dapat dibaca: {img_path}")

    img_original = cv2.cvtColor(
        img_bgr,
        cv2.COLOR_BGR2RGB
    )

    # Salinan gambar untuk hasil prediksi
    img_pred = img_original.copy()

    h, w = img_pred.shape[:2]

    # ========================================================
    # HITUNG GROUND TRUTH
    # ========================================================
    gt_counts = count_ground_truth(
        label_path,
        class_names
    )

    # ========================================================
    # PREDIKSI MODEL
    # ========================================================
    result = model.predict(
        img_path,
        conf=conf,
        iou=iou,
        verbose=False
    )[0]

    pred_counts = {
        name: 0
        for name in class_names
    }

    # Warna bounding box
    colors = {
        'table': (255, 80, 80),
        'person': (80, 200, 80)
    }

    # ========================================================
    # GAMBAR BOUNDING BOX DAN LABEL
    # ========================================================
    for box in result.boxes:

        cls_id = int(box.cls[0])

        if cls_id >= len(class_names):
            continue

        cls_name = class_names[cls_id]

        pred_counts[cls_name] += 1

        # Koordinat bounding box
        x1, y1, x2, y2 = map(
            int,
            box.xyxy[0].cpu().numpy()
        )

        conf_score = float(box.conf[0])

        color = colors.get(
            cls_name,
            (255, 255, 0)
        )

        # ====================================================
        # BOUNDING BOX TIPIS
        # ====================================================
        cv2.rectangle(
            img_pred,
            (x1, y1),
            (x2, y2),
            color,
            1
        )

        # ====================================================
        # LABEL KECIL TANPA BACKGROUND
        # ====================================================
        label_text = f"{cls_name} {conf_score:.2f}"

        font_scale = 0.25
        font_thickness = 1

        (tw, th), baseline = cv2.getTextSize(
            label_text,
            cv2.FONT_HERSHEY_SIMPLEX,
            font_scale,
            font_thickness
        )

        # Posisi label di atas bounding box
        text_x = x1
        text_y = y1 - 3

        # Jika terlalu dekat dengan batas atas
        if text_y - th < 0:
            text_y = y1 + th + 3

        # Jika keluar dari sisi kanan
        if text_x + tw > w:
            text_x = max(0, w - tw - 2)

        # Tulis label tanpa background
        cv2.putText(
            img_pred,
            label_text,
            (text_x, text_y),
            cv2.FONT_HERSHEY_SIMPLEX,
            font_scale,
            color,
            font_thickness,
            cv2.LINE_AA
        )

    return (
        img_pred,
        img_original,
        gt_counts,
        pred_counts
    )


# ============================================================
# PILIH 4 GAMBAR TEST SECARA ACAK
# ============================================================
N_SIMULASI = 4

valid_extensions = (
    '.jpg',
    '.jpeg',
    '.png',
    '.bmp'
)

all_test_files = [
    f for f in os.listdir(test_img_dir)
    if f.lower().endswith(valid_extensions)
]

sim_files = random.sample(
    all_test_files,
    min(N_SIMULASI, len(all_test_files))
)


# ============================================================
# SIAPKAN FIGURE
#
# BARIS 1 = HASIL PREDIKSI
# TENGAH   = INFORMASI SELISIH
# BARIS 2 = GAMBAR ASLI / GROUND TRUTH
# ============================================================

# Tinggi figure diperkecil agar ruang putih tidak terlalu besar
fig, axes = plt.subplots(
    2,
    4,
    figsize=(24, 8)
)

# Perkecil jarak horizontal dan vertikal antar gambar
plt.subplots_adjust(
    left=0.01,
    right=0.99,
    top=0.90,
    bottom=0.03,
    wspace=0.02,
    hspace=0.28
)


# ============================================================
# INISIALISASI TOTAL COUNTING
# ============================================================
all_gt = {
    name: 0
    for name in CLASS_NAMES
}

all_pred = {
    name: 0
    for name in CLASS_NAMES
}


# ============================================================
# PROSES SETIAP GAMBAR
# ============================================================
for i, fname in enumerate(sim_files):

    img_path = os.path.join(
        test_img_dir,
        fname
    )

    label_path = os.path.join(
        test_labels_dir,
        os.path.splitext(fname)[0] + '.txt'
    )

    (
        vis_img,
        original_img,
        gt_counts,
        pred_counts
    ) = simulate_counting_visual(
        best_model,
        img_path,
        label_path,
        CLASS_NAMES,
        CONF_THRESHOLD,
        IOU_THRESHOLD
    )

    # ========================================================
    # AKUMULASI COUNTING
    # ========================================================
    for name in CLASS_NAMES:
        all_gt[name] += gt_counts[name]
        all_pred[name] += pred_counts[name]


    # ========================================================
    # HITUNG SELISIH PER GAMBAR
    # ========================================================
    differences = {
        name: abs(
            gt_counts[name] - pred_counts[name]
        )
        for name in CLASS_NAMES
    }


    # ========================================================
    # BARIS PERTAMA
    # HASIL PREDIKSI
    # ========================================================
    axes[0, i].imshow(vis_img)

    counting_text = " | ".join(
        [
            f"{name}: {pred_counts[name]}"
            for name in CLASS_NAMES
        ]
    )

    axes[0, i].set_title(
        f"Hasil Prediksi\n{counting_text}",
        fontsize=10,
        fontweight='bold',
        pad=3
    )

    axes[0, i].axis('off')


    # ========================================================
    # BARIS KEDUA
    # GAMBAR ASLI / GROUND TRUTH
    # ========================================================
    axes[1, i].imshow(original_img)

    gt_text = " | ".join(
        [
            f"{name}: {gt_counts[name]}"
            for name in CLASS_NAMES
        ]
    )

    axes[1, i].set_title(
        f"Gambar Asli / Ground Truth\n{gt_text}",
        fontsize=10,
        pad=3
    )

    axes[1, i].axis('off')


    # ========================================================
    # INFORMASI SELISIH DI ANTARA DUA BARIS GAMBAR
    # ========================================================
    diff_text = " | ".join(
        [
            f"Selisih {name}: {differences[name]}"
            for name in CLASS_NAMES
        ]
    )

    # Ambil posisi gambar baris atas dan bawah
    top_pos = axes[0, i].get_position()
    bottom_pos = axes[1, i].get_position()

    # Posisi tengah antara gambar prediksi dan gambar asli
    center_x = (
        top_pos.x0 + top_pos.x1
    ) / 2

    center_y = (
        top_pos.y0 + bottom_pos.y1
    ) / 2

    # Tampilkan informasi selisih
    fig.text(
        center_x,
        center_y,
        diff_text,
        ha='center',
        va='center',
        fontsize=9,
        fontweight='bold'
    )


# ============================================================
# MATIKAN AXIS JIKA GAMBAR KURANG DARI 4
# ============================================================
for i in range(len(sim_files), 4):
    axes[0, i].axis('off')
    axes[1, i].axis('off')


# ============================================================
# JUDUL UTAMA
# ============================================================
fig.suptitle(
    "Perbandingan Hasil Object Counting dengan Ground Truth",
    fontsize=15,
    fontweight='bold',
    y=0.97
)


# ============================================================
# TAMPILKAN HASIL
# ============================================================
plt.show()


# ============================================================
# RINGKASAN SIMULASI COUNTING
# ============================================================
print("\n========================================")
print("RINGKASAN SIMULASI OBJECT COUNTING")
print("========================================")

for name in CLASS_NAMES:

    diff = abs(
        all_gt[name] - all_pred[name]
    )

    print(
        f"{name:8s}: "
        f"Ground Truth = {all_gt[name]:3d} | "
        f"Prediksi = {all_pred[name]:3d} | "
        f"Selisih = {diff}"
    )
```

## 15. FPS (Kecepatan Inferensi)

Threshold confidence & IoU memakai default Ultralytics (conf=0.25, iou=0.7) — tidak ada threshold tuning di baseline ini.


```python
test_img_dir = os.path.join(TEST_DIR, 'images')
image_files = sorted(glob.glob(os.path.join(test_img_dir, '*.*')))
sample_imgs = image_files[:30] if len(image_files) >= 30 else image_files

CONF_THRESHOLD = 0.3
IOU_THRESHOLD = 0.5

# Warm-up
_ = best_model.predict(sample_imgs[0], conf=CONF_THRESHOLD, iou=IOU_THRESHOLD, verbose=False)

times = []
for img_path in sample_imgs:
    start = time.time()
    _ = best_model.predict(img_path, conf=CONF_THRESHOLD, iou=IOU_THRESHOLD, verbose=False)
    times.append(time.time() - start)

avg_time = np.mean(times)
fps = 1 / avg_time

print(f"Rata-rata waktu inferensi : {avg_time * 1000:.2f} ms/gambar")
print(f"FPS                       : {fps:.2f}")
```

## 16. Object Counting — MAE


```python
def count_ground_truth(label_path, class_names):
    counts = {name: 0 for name in class_names}
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                line = line.strip()
                if not line:
                    continue
                cls_id = int(line.split()[0])
                counts[class_names[cls_id]] += 1
    return counts

test_labels_dir = os.path.join(TEST_DIR, 'labels')

results_list = []
for img_path in image_files:
    filename = os.path.basename(img_path)
    label_path = os.path.join(test_labels_dir, os.path.splitext(filename)[0] + '.txt')

    gt_counts = count_ground_truth(label_path, CLASS_NAMES)

    pred = best_model.predict(img_path, conf=CONF_THRESHOLD, iou=IOU_THRESHOLD, verbose=False)[0]
    pred_counts = {name: 0 for name in CLASS_NAMES}
    for box in pred.boxes:
        cls_id = int(box.cls[0])
        pred_counts[CLASS_NAMES[cls_id]] += 1

    row = {'filename': filename}
    for name in CLASS_NAMES:
        row[f'gt_{name}'] = gt_counts[name]
        row[f'pred_{name}'] = pred_counts[name]
    results_list.append(row)

df_counting = pd.DataFrame(results_list)

print("=== MAE Counting per Kelas (Baseline) ===")
mae_rows = []
for name in CLASS_NAMES:
    gt = df_counting[f'gt_{name}']
    pred = df_counting[f'pred_{name}']
    mae = np.mean(np.abs(gt - pred))
    mae_rows.append({'class': name, 'MAE': mae})
    print(f"Kelas '{name}': MAE = {mae:.3f}")

df_mae_summary = pd.DataFrame(mae_rows)

df_counting.to_csv(os.path.join(PROJECT_DIR, 'counting_results_yolov8n_baseline_per_image.csv'), index=False)
df_mae_summary.to_csv(os.path.join(PROJECT_DIR, 'counting_results_yolov8n_baseline_mae_summary.csv'), index=False)

df_mae_summary
```
